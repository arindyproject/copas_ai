```python
import feedparser
from datetime import datetime
from urllib.parse import quote_plus
# Query kata kunci
query = "geo politik"
url = f"https://news.google.com/rss/search?q={quote_plus(query)}&hl=id&gl=ID&ceid=ID:id"

# Parsing RSS feed
feed = feedparser.parse(url)

print(f"📢 Total berita: {len(feed.entries)}\n")

for entry in feed.entries:
    # Format tanggal
    if hasattr(entry, "published_parsed"):
        published_dt = datetime(*entry.published_parsed[:6])
        published_str = published_dt.strftime("%Y-%m-%d %H:%M:%S")
    else:
        published_str = "Unknown"

    # Sumber berita dan link sumber
    source_name = entry.source.title if hasattr(entry, "source") and hasattr(entry.source, "title") else "Tidak diketahui"
    source_link = entry.source.href if hasattr(entry, "source") and hasattr(entry.source, "href") else "Tidak tersedia"

    print("📰 Judul        :", entry.title)
    print("🏢 Sumber       :", source_name)
    print("🔗 Link Sumber  :", source_link)
    print("📅 Tanggal      :", published_str)
    print("🌐 Link Berita  :", entry.link)
    print("-" * 110)
```

```python
import requests
from bs4 import BeautifulSoup

# URL yang mau diambil
url = "https://www.cnbcindonesia.com/research/20250618141632-128-641981/sri-mulyani-beberkan-geopolitik-rugikan-as-ini-buktinya"  # contoh: halaman utama Kompas

# Kirim permintaan ke situs
response = requests.get(url)

# Ubah isi HTML ke objek BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# Ambil semua judul berita (misal di tag <h3>)
titles = soup.find_all("p")

# Tampilkan hasil
for i, t in enumerate(titles[:5]):  # tampilkan 5 judul pertama
    print(f"{i+1}.", t.get_text(strip=True))
```
